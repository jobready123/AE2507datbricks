```python
%sql
create catalog if not exists DBS0808;
use catalog DBS0808;
create schema if not exists schema_demo;
use schema_demo;

create volume if not exists data_volume;
```

```python
dbutils.fs.mkdirs("/Volumes/dbs0808/schema_demo/data_volume/d_1")
```

```python
data = [
    ("1", "a"),
    ("2", "b"),
    ("3", "c")  
]
df = spark.createDataFrame(data, ["id", "value"])
df.show()
df.display()

output_path = "/Volumes/dbs0808/schema_demo/data_volume/f_1"
df.write.mode("overwrite").format("csv").save(output_path)

output_path = "/Volumes/dbs0808/schema_demo/data_volume/d_1"
df.write.mode("overwrite").format("delta").save(output_path)
```
