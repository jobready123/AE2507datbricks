```python
%sql
use catalog jobready123;
use schema default;
CREATE VOLUME my_volume
COMMENT 'My storage volume for demo files';

SHOW VOLUMES
```

```python
dbutils.fs.mkdirs("/Volumes/jobready123/default/my_volume/silver")
```

```python


print('test2')
# Make sure Spark is imported
from pyspark.sql import SparkSession

# In Databricks, `spark` is already defined, but this ensures we have it
spark = SparkSession.builder.getOrCreate()

data = [
    (1, "Alice", 5000),
    (2, "Bob", 4000),
    (3, "Charlie", 3000)
]
df = spark.createDataFrame(data, ["id", "name", "salary"])
df.display()

# Path inside your Volume

output_path = "/Volumes/jobready123/default/my_volume/silver/output_csv"

# Save as CSV
df.write.mode("overwrite").option("header", "true").csv(output_path)

print(f"Data written to: {output_path}")
```
