```python
from pyspark.sql import SparkSession

from pyspark.ml.feature import StringIndexer as PyStringIndexer
spark = SparkSession.builder.getOrCreate()
StringIndexer = PyStringIndexer

# Sample data
df = spark.createDataFrame([
    ("Canada",),
    ("USA",),
    ("Mexico",),
    ("Canada",),
], ["Country"])

# Use safe parameter setting
indexer = StringIndexer() \
    .setInputCol("Country") \
    .setOutputCol("Country_Index")

model = indexer.fit(df)
df_indexed = model.transform(df)

df_indexed.show()
```
