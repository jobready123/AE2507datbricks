```python
%sql
use catalog 0725catalog;
drop table if exists 0725catalog.bronze.customers_raw;
CREATE TABLE IF NOT EXISTS 0725catalog.bronze.customers_raw (
  customerID STRING,
  country STRING
) USING DELTA;

drop table if exists 0725catalog.silver.customers_dim;
CREATE TABLE IF NOT EXISTS 0725catalog.silver.customers_dim (
  customerID STRING,
  country STRING,
  start_date DATE,  -- When this version became active
  end_date DATE,    -- When it was superseded (null if current)
  current BOOLEAN   -- Boolean flag for active recor
)
USING DELTA;
```
